## Data Pre-processing:

Data pre-processing is the most important step because if we do not clean and structured data, we cannot get actual results. We will find empty values, invalid entries etc. in this step.

```python
healthcare.isnull().sum()
```

The isnull() is used to check the missing values. In out dataset column SelfEmployed has 4356 null entries and we cannot assume these entries and take an average so will remove these entries. Our dataset is quite big so it will hurt the final analysis.

```python
healthcare = healthcare.dropna(subset=['SelfEmployed'])
```

```python
healthcare.isnull().sum()
```

No more missing entries. The .dropna() function check the given column entries if the value is missing it drops that entire row.

```python
healthcare.duplicated().sum()
```

In python pandas .duplicated() function returns the values that are exactly duplicated with other values. In our dataset we 174897 duplicated entries so we will remove these values to get the accurate results. This removal will leave us with substantial clean dataset.

```python
healthcare = healthcare.drop_duplicates()
```

```python
healthcare.shape
```

The .drop_duplication() function drop or deletes the duplicated entries as in our case we had 174897 enties so after deleting this number we get 82017 rows which previously were 261328. As we can see from our columns we SocialWeakness and SocialWeakness.1 columns that looks similar. Lets find out if they have exact same entries.

```python
print(
    (healthcare['SocialWeakness'] == healthcare['SocialWeakness.1']).all()
)
```

SocialWeakness and SocialWeakness.1 have the same entries so we can delete SocialWeakness.1 column as it does not serve any purpose.

```python
healthcare = healthcare.drop(columns=['SocialWeakness.1'])
healthcare.head()
```

The output shows that SocialWeakness.1 column has been removed successfully. 

```python
print(healthcare.dtypes)
```

## Insights:

### Insight 1: How does Family History is affecting Treatment?

```python
family_treatment = pd.crosstab(
    healthcare['FamilyHistory'],
    healthcare['Treatment'],
    normalize='index'
) * 100

ax = family_treatment.plot(
    kind='bar', stacked=True, figsize=(7,5)
)

plt.title('Treatment Status by Family History')
plt.xlabel('Family History')
plt.ylabel('Percentage of Respondents')
plt.legend(title='Treatment')
plt.show()
```

The chart shows that the patients with family history have a higher treatment proportion. This finding helps the organisation to understand that employee mental health can differ according to family background. For technical concepts, the crosstab() function in pandas compares the 2 columns of the same table or dataset. Then, I have used simple bar chart to show how family history can affect on getting treatment.

### Insight 2: Does Gender and Treatment have any relation?

```python
gender_treatment = pd.crosstab(
    healthcare['Gender'],
    healthcare['Treatment'],
    normalize='index'
) * 100

gender_treatment['Yes'].plot(
    kind='bar', figsize=(7,5)
)

plt.title('Treatment Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Treatment Rate (%)')
plt.ylim(0,100)
plt.show()
```

Female respondents are prone to higher treatment rate than male respondents.

### Insight 3: Does Occupation has any affects on Increasing Stress?

```python
occupation_stress = pd.crosstab(
    healthcare['Occupation'],
    healthcare['IncreasingStress'],
    normalize='index'
) * 100

print(occupation_stress.round(2))
```

The data shows that people in different occupations have different increasing stress rate like Business and Corporate has high rate of increasing stress. This relation will help in monitoring what occupation is more stressfull. Lets see these finding in a graph for better understanding.

```python
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.heatmap(
    occupation_stress,
    annot=True,
    fmt='.3f',
    cmap='Reds',
    vmin=0,
    vmax=100
)

plt.title('Increasing Stress Across Occupation Groups')
plt.xlabel('Increasing Stress')
plt.ylabel('Occupation')

plt.show()
```

The heatmap shows that Business has highest increasing stress rate and Student has lowest increasing stress rate. For technical concepts, I have imported Seaborn Library to draw heatmap(Waskom, M. (2026)). The heatmap function of Seaborn library takes multiple arguments, occupation stress is the calculation that we have done in previous step, annote shows the values in numbers, fmt decides till how many decimal we want to show e.g if I want to show till 1 decimal I will use .1f, cmap decides the colors of the heatmap, vmin and vmax are the minimum starting values and maximum ending values respectively.

### Insight 4: Does staying indoors for longer periods affects copping struggles?

```python
days_coping = pd.crosstab(
    healthcare['DaysIndoors'],
    healthcare['CopingStruggles'],
    normalize='index'
) * 100

print(days_coping.round(2))
```

Comparitively, the more time you spend indoors the higher copingstruggles rate is. Let me show the findings in a graph for better understanding.

```python
coping_rate = days_coping['Yes']

plt.figure(figsize=(8, 5))
plt.plot(coping_rate.index, coping_rate.values, marker='o')
plt.title('Coping Struggles by Time Spent Indoors')
plt.xlabel('Days Indoors')
plt.ylabel('Coping Struggles (%)')
plt.ylim(0, 60)
plt.grid(True, alpha=0.2)
plt.show()
```

The chart shows that more number of days have higher coping struggle percentage. Although the longest time duration falls slightly lead by Go out everyday

### Insight 5: Does gender affects in opening up for mental health interviews?

```python
gender_interview = pd.crosstab(
    healthcare['Gender'],
    healthcare['MentalHealthInterview'],
    normalize='index'
) * 100

print(gender_interview.round(2))
```

The outputs tells that Females have higher rate of saying No for mental health interviews compared to Male respondents. Lets see this in a chart for clear understanding:

```python
gender_interview.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Mental Health Interview Responses by Gender')
plt.xlabel('Gender')
plt.ylabel('Percentage of Respondents')
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(title='Response')
plt.tight_layout()
plt.show()
```

This chart clears the concept that more female said No for mental health interview while Most male respondents said Yes or Maybe for the same task. 

### Insight 6:  How do respondents percieve the availability of mental health care options?

```python
care_counts = healthcare['CareOptions'].value_counts()

print(care_counts)
print((care_counts / len(healthcare) * 100).round(2))
```

This is very interesting finding as only 31.39% respondent said they knew about available care options. The other 68.71% respondent either were not sure or did not know the available care options. This finding helps organisations decide that they to work on awareness of available care options because majority of the people does not even know that there are mental health care options available.

```python
plt.pie(
    care_counts.values,
    labels=care_counts.index,
    autopct='%1.2f%%',
    startangle=90,
    colors=['#ff3b3b','#3fcb3b', '#1a90e1'] #No = Softred, Yes=SoftGreen, NotSure=Softblue
)

plt.title('Perceived Availability of Mental Health Care Options')
plt.show()
```

Pie chart shows the distribution of responses to the mental health care otions availability.

### Insight 7: What countries have high number of respondents?

```python
country_counts = healthcare['Country'].value_counts()

print(country_counts.head(10))
print((country_counts.head(10) / len(healthcare) * 100).round(2))
```

```python
top_countries = country_counts.head(10).sort_values()

plt.figure(figsize=(8, 5))
plt.barh(top_countries.index, top_countries.values)

plt.title('Top 10 Countries by Number of Respondents')
plt.xlabel('Number of Respondents')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

We have selected top 10 countries with the highest number of respondent. From those top 10 only USA and UK make the most of number of respondents. I have used horizontal bar graph to show the top 10 countries in Ascending order on the bases of number of respondents from each country. This data tells the organisations that what respondents from which countries are more willing to participate in data gathering.
